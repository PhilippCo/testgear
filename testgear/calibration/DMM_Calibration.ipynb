## DUT

```python
dut = testgear.base_classes.instrument(ip="172.16.0.120")
#dut = testgear.base_classes.instrument(gwip="172.16.0.27", gpib=5)

dmm  = dut.query("*IDN?")
visa = dut.visastr
dut.close()

dmm.split(',')[1]
```

### choose correct class

```python
if dmm.split(',')[1] == "34470A":
    dut = testgear.HPAK.K34470A(visa)
    
elif dmm.split(',')[1] == "34461A":
    dut = testgear.HPAK.K34461A(visa)

elif dmm.split(',')[1] == "34465A":
    dut = testgear.HPAK.K34465A(visa)

elif dmm.split(',')[1] == "8588A":
    dut = testgear.Fluke.F8588A(visa)

elif dmm.split(',')[1] == "34401A":
    dut = testgear.HPAK.HP34401A(visa)

elif dmm.split(',')[1] == "MODEL DMM6500":
    dut = testgear.Keithley.K6500(visa)
    
else:
    raise SystemExit("DMM not supported!")

```

## Calibrator

```python
calibrator = testgear.Fluke.F5730A(gwip="172.16.0.27", gpib=4)
calibrator
```

```python
calibration.calibrate(calibrator, dut)
```
